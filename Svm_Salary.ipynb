```python

import pandas as pd
train_data = pd.read_csv("SalaryData_Train.csv")
test_data = pd.read_csv("SalaryData_Test.csv")

train_data.duplicated().sum()
train_data = train_data.drop_duplicates()
train_data
test_data.duplicated().sum()
test_data=test_data.drop_duplicates()
test_data
```

```python
# Assuming 'Salary' is the target variable
X_train = train_data.drop('Salary', axis=1)
y_train = train_data['Salary']

X_test = test_data.drop('Salary', axis=1)
y_test = test_data['Salary']
```

```python
# Identify numeric and categorical features
numeric_features = X_train.select_dtypes(include=['int64', 'float64']).columns
categorical_features = X_train.select_dtypes(include=['object']).columns
```

```python

# Create transformers for numeric and categorical features
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
numeric_transformer = StandardScaler()
categorical_transformer = Pipeline(steps=[
    ('label_encoder', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
# Create preprocessor to handle different types of features
from sklearn.compose import ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])
```

```python
# Create an SVM pipeline
from sklearn.svm import SVC
svm_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', SVC(kernel='linear'))
])
```

```python
svm_pipeline.fit(X_train, y_train)
```

```python
y_pred = svm_pipeline.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
classification_rep = classification_report(y_test, y_pred)
```

```python
print('Accuracy:',accuracy)
print('Confusion Matrix:',conf_matrix)
print('Classification Report:',classification_rep)
```

```python
# Create an SVM pipeline for rbf method
from sklearn.svm import SVC
svm_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', SVC(kernel='rbf'))
])

# Train the SVM model
svm_pipeline.fit(X_train, y_train)

# Make predictions on the test set
y_pred = svm_pipeline.predict(X_test)

# Evaluate the model
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
classification_rep = classification_report(y_test, y_pred)

print('Accuracy:',accuracy)
print('Confusion Matrix:',conf_matrix)
print('Classification Report:',classification_rep)
```

```python
# Create an SVM pipeline for poly method
from sklearn.svm import SVC
svm_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', SVC(kernel='poly'))
])

# Train the SVM model
svm_pipeline.fit(X_train, y_train)

# Make predictions on the test set
y_pred = svm_pipeline.predict(X_test)

# Evaluate the model
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
classification_rep = classification_report(y_test, y_pred)

print('Accuracy:',accuracy)
print('Confusion Matrix:',conf_matrix)
print('Classification Report:',classification_rep)
```
